# **Introduction**


## **Project Overview**


Music is fundamentally driven by emotion, and lyrics provide a window straight to the artist's mind, where thoughts happen and stories are made. Different musical genres represent feelings differently, and explore various thematics. Through textual patterns, there's an opportunity to map emotional landscapes. This project uses Natural Language Processing (NLP) and machine learning to decode these lyrical structures and uncover relationships between words, genres and emotions.


## **Main goals**

This project tackles 2 primary challenges:

 - Genre Identification: we aim to build a robust classification model that's capable of predicting a track's genre purely through its lyrics. This challenges the model to identify the patterns associated with different musical styles.

 - Sentiment Analysis: we will conduct an exploratory sentiment analysis to quantify the emotional weight of each song's lyrics. By generating and evaluating sentiment scores, we will investigate how emotional trends vary between genres, track how predominant sentiments evolve over time, and explore whether a song's emotional polarity correlates with its overall popularity.
